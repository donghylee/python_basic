## 7. 문자열 포매팅

### 1) 숫자 바로 대입

```python
"I eat %d apples."%3
```

```python
"I eat %d apples."%5
```

### 2) 문자열 바로 대입

```python
"I eat %d apples."%five
```

```python
"I eat %d apples."%"five"
# %d는 숫자만 s로 변경 후 입력하고 싶은 단어에 ""를 묶어서 작성해주면 된다.
```

```python
"I eat %s apples."%"five"
```

### 3) 숫자 값을 나타내는 변수로 대입

```python
num = 3
"I eat %d apples."%num
#num <에 왜 ""로 안묶고 사용하는가? -> 숫자를 포함한 변수로 만들었기 때문에
```

### 4) 2개 이상의 값 넣기

```python
number = 10
day = 'three'

"I ate %d apples, so i was sick for %s days"%(number, day)
#2개 이상의 값을 넣으려면 마지막 % 다음 괄호 안에 쉼표(,)로 구분하여 각각의 값을 넣어 주면 된다.
#()안에 배치는 %의 순서대로 작성한다. 예) (day, number)로 하면 에러가 발생함. -> %d=숫자, %s=문자 이렇게 순서대로 되어있기 때문에.
```

## 9. 문자열 포맷 코드

```python
# %s는 숫자도 문자열로도 대입 가능.
"I have %s apples" %3
```

```python
"rate is %s" % 3.234
```

```python
"rate is %d" % 3.234
# %d = 정수만 표현되기 때문에 소수점 뒤의 자리는 다 삭제 된다.
```

```python
"rate is %f" % 3.234
# %f 는 부동소수 인데 기본 세팅값이 소수점 뒤 6자리 고정.
```

```python
# 포맷 코드와 문자(%)를 같이 사용하려면 %%(두번 사용)하면 된다.
"Error is %d%%." %98
```

```python
# 뒤에 % 하나면 그 뒤에 문자나 숫자열이 들어와야 하는데 아무것도 안들어 왔으므로 에러가 발생.
"Error is %d%." %98
```

```python
# 포맷 코드를 작성 안하고 하면 크게 상관 없음.
"Error is 98%."
```

### 포맷 코드와 숫자 함께 사용하기

```python
# %10<- 문장의 전체 길이를 표시함. 그러므로 공백8자리를 포함한 hi의 입력값이 나옴.
"%10s" %"hi"
```

```python
a = "%10s" %"hi"
a
```

```python
len(a)
```

```python
a[-2]
```

```python
# -를 붙이면 입력값을 가장 앞으로 표시하고 그 뒤에 공백을 추가해서 출력 됨.
a = "%-10s" %"hi"
a
```

```python
"%-10sjane."%'hi'
```

```python
# 소수점(.) 뒤에 이어지는 숫자(4)는 그 자리까지만 출력을 한다는 의미다.
"%0.4f"%3.42134234
```

```python
# %(10)은 문자열 전체 길이를 10칸까지 표기한다는 의미.
b = "%10.4f" %3.42134234
b
```

```python
len(b)
```

```python
"%-10.4f" %3.42134234
```

## .format()함수를 사용한 포매팅

```python
"I eat {} apples".format(3)
```

```python
"I eat {} apples".format('six')
```

```python
num = 3
"I eat {} apples".format(num)
```

```python
number = 10
day = "three"
"I ate {0} apples. so i was sick for {1} days.".format(number, day)
```

```python
# 이 경우는 순서대로 작성을 할 필요가 없다. .format(day="", nember="")로 해도 출력값은 그대로임.
"I ate {number} apples. so i was sick for {day} days.".format(number=10, day="three")
```

```python
# 인덱스와 변수명을 혼용해서 넣기. (이럴 경우에는 순서대로 작성을 해줘야 함.)
"I ate {0} apples. so i was sick for {day} days.".format(10, day="three")
```

```python
# :< 왼쪽 정렬 / :> 오른쪽 정렬 / :^ 가운대 정렬
"{:<10}".format("hi")
```

```python
"{:>10}".format("hi")
```

```python
"{:^10}".format("hi")
```

```python
# 공백 채우기
"{:=^10}".format("hi")
```

```python
"{:!<10}".format("hi")
```

```python
# 소수점 표현
y = 3.42134234
"{:=>10.4f}".format(y)
```

```python
"{{and}}".format()
```
