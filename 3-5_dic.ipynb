딕셔너리 만들때 주의사항

```python
a = {1:'a', 1:'b'}  #키 값은 고유한 값이다.(중복 불가)
a
```

```python
a = {[1,2]:'hi'}   #key 안에는 리스트 작성 불가
```

```python
a = {(1,2):'hi'}   # key 안에 튜플 사용 가능.
a
```

```python
a = {1,2: 'hi'}
a
```

```python
a[(1,2)]
```

## 5. 딕셔너리 관련 함수

key 리스트 만들기 : keys()

```python
a = {'name': 'ldh', 'phone':'010-9540-0001', 'birth':'0401'}
```

```python
a.keys()
```

```python
type(a.keys())
```

```python
list(a.keys())
```

```python
type(list(a.keys()))
```

```python
for k in a.keys():
    print(k)
```

value 리스트 만들기 :values()

```python
a.values()
```

key, value 쌍 얻기 : items()

```python
a.items()
```

key:value 쌍 모두 지우기 : clear()

```python
a.clear()
a
```

key로 value 얻기 : get()

```python
a = {'name': 'ldh', 'phone':'010-9540-0001', 'birth':'0401'}

a.get('phone')
```

```python
print(a.get('age'))
```

```python
print(a['age'])
```

```python
    # 딕셔너리 안에 찾는 key가 없으면 디폴트 값을 가져오게 할 수 있다.
    # get(key,'default value')

a.get('age', 25)
```

```python
a
```

```python
a.get('phone', 35)
```
