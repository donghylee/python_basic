## 4. 리스트 관련 함수

리스트에 요소 추가 : append(value)

```python
a = [1,2,3]
a.append('four')
a
```

```python
a.append([5,6])
a
```

리스트 정렬 : sort()

```python
a = [1,4,3,2]
a.sort()
a
```

```python
a = ['d','f','z','x','w','a','d','b']
a.sort()
a
```

```python
a.sort(reverse=True)
a
```

리스트 뒤집기(인덱스 역순 정렬) : reverse()

```python
a = ['f','z','x','w','a','d','b']
a.reverse()
a
```

인덱스 변환 : index(value)

```python
a
```

```python
a.index('w')
```

```python
a.index('e')
```

리스트에 요소 삽입 : insert(index, value)

```python
a = [1,2,3]
a.insert(0,4)
a
```

```python
a.insert(3,5)
a
```

리스트 요소 제거 : remove(value)

```python
a = ['b','a','c','d']

a.remove('a')
a
```

```python
a = ['b','a','c','d','a','c','d']

a.remove('a')
a
```

리스트 요소 끄집어 내기 : pop(index)

```python
a = [1,2,3]

a.pop()
```

```python
a
```

```python
a.pop(0)
```

```python
a
```

리스트에 포함된 요소의 개수 세기 : count(value)

```python
a = [1,2,3,4,2,2,3,2]
a.count(2)
```

리스트 확장 : extend(value)

```python
a = [1,2,3,4]
a.extend([5,6])
a
```

```python
a = [1,2,3]
a.extend((3,[4,5]))
a
```

```python
# a = a + [4,5] 와 같은 의미
a = [1,2,3]
a += [4,5] 
a
```
