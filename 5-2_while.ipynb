```python
# tree_hit = 스네이크 방식 / TreeHit = 파스칼 케이스 / treeHit = 카멜 케이스

treeHit = 0
while treeHit < 10:
    treeHit += 1
    print("나무를 %d번 찍었습니다."%treeHit)
    if treeHit==10:
        print("나무가 넘어갑니다.")
```

### break문

```python
    # 커피 자판기

coffee = 10
money = 300

while money:
    print("돈을 받았으니 커피를 줍니다.")
    coffee -= 1
    print("남은 커피의 양은 %d개 입니다."%coffee)
    if coffee== 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다.")
        break
```

```python
coffee = 10

while True:
    money = int(input("돈을 넣어 주세요> "))
    if money == 300:
        print("커피가 나옵니다.")
        print("남은 커피의 양은 %d개 입니다."%coffee)
        coffee -= 1
    elif money > 300:
        print("거스름돈 %d를 주고 커피가 나옵니다."%(money-300))
        print("남은 커피의 양은 %d개 입니다."%coffee)
        coffee -= 1
    else:
        print("돈을 다시 돌려주고 커피가 안나옵니다.")
        print("남은 커피의 양은 %d개 입니다."%coffee)
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지 합니다.")
        break
```

### continue문

```python
a = 0
while a <10:
    a += 1
    if a % 2 == 0: continue
    print(a)
```

총 5명의 학생이 시험을 보았는데 시험 점수가 60점 이상이면 합격이고 그렇지 않으면 불합격이다. 합격인지, 불합격인지 결과를 보여 주시오.

```python
marks = [90, 25, 67, 45, 80]    # 학생들의 시험 점수 리스트

num = 0  # 학생에게 붙여 줄 번호

for mark in marks:      # 90,25,67,45,80 을 순서대로 mark에 대입
    num += 1
    if mark >= 60:
        print("%d번 학생은 합격입니다."%num)
    else:
        print("%d번 학생은 불합격입니다"%num)
```

```python
marks = [90, 25, 67, 45, 80]

num = 0

for mark in marks:
    num += 1
    if mark < 60: continue
    print(f"{num}번 학생 축하합니다. 합격입니다.")
```

### 리스트 컴프리헨션(list comprehension)

```python
    # 일반적인 for문
a = [1,2,3,4]
result = []

for num in a:
    result.append(num*3)
print(result)
```

```python
result = [num*5 for num in a]
print(result)
```

```python
    # 리스트 컴프리헨션안에 if문
result = [num*2 for num in a if num%2 == 0]
print(result)
```

```python
    # 리스트 컴프리헨션으로 구구단
result = [x*y for x in range(2,10)
            for y in range(1,10)]
print(result)
```
