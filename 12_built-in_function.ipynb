filter(함수, 반복가능한 데이터) : 반복가능한 데이터를 요소 순서대로 함수를 호출 했을때 참인것만 리턴

```python
    # 일반 함수로 양수값만 리턴
def positive(li):
    result = []
    for i in li:
        if i > 0:
            result.append(i)
    return result

positive([1,-3,2,0,-5,6])
```

```python
    # 필터 함수로 양수값만 리턴
def positive(x):
    return x > 0

list(filter(positive, [1,-3,2,0,-5,6]))
```

```python
    # lambda 함수로 양수값만 리턴
list(filter(lambda x: x>0,[1,-3,2,0,-5,6]))
```

map(함수, 반복가능한 데이터) : 반복 가능한 데이터를 입력받아 각 요소별로 함수를 적용한 결과를 리턴

```python
    # 일반 함수로 리스트의 각 요소에 2를 곱한 값을 리턴.
def two_times(numberList):
    result = []
    for number in numberList:
        result.append(number*2)
    return result

two_times([1,2,3,4])
```

```python
    # map 함수로 리스트에 각 요소에 2를 곱한 값을 리턴
def two_times(x):
    return x*2

list(map(two_times,[1,2,3,4]))
```

```python
list(map(lambda x: x*2, [1,2,3,4]))
```
