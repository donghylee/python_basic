## 2. 자료형의 참과 거짓

자료형의 참과 거짓이 프로그램에서 어떻게 쓰이는지 간단히 알아보자.

```python
    # a.pop() : 리스트를 빼내는 기능으로 모든 리스트를 빼내고 나서 더 이상 뺄게 없으므로 작동 정지가 되어서 4를 끝으로 멈춘 것.
a = [1,2,3,4]

while a:
    print(a.pop())
```

```python
    # 리스트에 값이 없으면 거짓으로 출력 된다.
if [1,2]:
    print('참')
else:
    print('거짓')
```

```python
if []:
    print('참')
else:
    print('거짓')
```
