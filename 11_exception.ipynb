## 1. 예외 처리 기법

try-except 구문

```python
try:
    4/0
except ZeroDivisionError as e:
    print(e)
```

```python
try:
    print(4/3)
except ZeroDivisionError as e:
    print(e)
```

try-finally 문

```python
try:
    f = open('non.txt','r')
except FileNotFoundError as e:
    print(e)
finally:
    f.close()
```

```python
try:
    f = open('foo.txt','w')
    # 예외발생하면 이후에 실행문 실행 안됨!
    f.close()
finally:
    f.close()
```

여러개 예외 처리

try-else 문

```python
try:
    age=int(input('나이를 입력하세요: '))
except:
    print('입력이 정확하지 않습니다.')
else:
    if age <= 18:
        print('미성년자는 출입금지입니다.')
    else:
        print('환영합니다.')
```

## 3. 예외 발생

```python
class Bird:
    def fly(self):
        raise NotImplementedError
```

```python
class Eagle(Bird):
    pass

eagle = Eagle()
eagle.fly()
```

```python
class Eagle(Bird):
    def fly(self): # 메서드 오버라이딩
        print('very fast')
eagle = Eagle()
eagle.fly()
```

## 4. 예외 만들기

```python
class MyError(Exception):
    pass
```

```python
def say_nick(nick):
    if nick =='바보':
        raise MyError # 강제로 내가 만든 예외 발생
    print(nick)
```

```python
say_nick("천사")
```

```python
say_nick("바보")
```

```python
try:
    say_nick('천사')
    say_nick('바보')
except MyError:
    print('허용되지 않는 별명입니다.')
```

```python
try:
    say_nick('천사')
    say_nick('바보')
except MyError as e:
    print(e)
```

```python
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."
```

```python
try:
    say_nick('천사')
    say_nick('바보')
except MyError as e:
    print(e)
```
