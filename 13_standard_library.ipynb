datetime.date

A 군과 B 양이 2024년 12월 14일부터 만나기 시작했다면 2026년 4월 5일은 둘이 사귄 지 며칠째 되는 날일까?

아울러 사귀기 시작한 2024년 12월 14일은 무슨 요일이었을까?

datetime.date 함수를 사용하면 이 문제를 쉽게 해결할 수 있다.

```python
import datetime

day1 = datetime.date(2024,12,14)
day2 = datetime.date(2026,4,5)

diff = day2 - day1
diff
```

### random : 난수(규칙이 없는 임의의 수)를 발생시키는 모듈

```python
import random

random.random()
```

```python
    # randint() : 정수 중에서 난수 발생
random.randint(1,10)
```

```python
    # random_pop 함수
def random_pop(data):
    num = random.randint(0, len(data)-1)
    print(data, num)
    return data.pop(num)

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

random.choice() : 입력받은 리스트에서 무작위로 하나의 요소를 선택하여 리턴

```python
def random_pop(data):
    num = random.choice(data)
    print(data, num)
    data.remove(num)
    return num

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

random.sample() : 리스트 요소를 무작위로 섞어서 반환해

```python
data = (1,2,3,4,5)
random.sample(data, len(data))
```

pickle : 객체의 형태를 그대로 유지하면서 파일을 저장하고 불러오게 하는 모듈(피클 데이터는 바이너리 데이터)

```python
import pickle

with open('test.txt', 'wb') as f:
    data = {1: 'python', 2: 'you need'}
    pickle.dump(data, f)
```

```python
with open('test.txt', 'rb') as f:
    content = pickle.load(f)
    print(content)
```

json : json 데이터를 처리할 때 사용

```python
import json

with open('myinfo.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
    # dict 자료형을 -> json 파일로
data = {'name': '김경일', 'birth': '0411', 'age': 35}

with open('myinfo.json', 'w') as f:
    json.dump(data,f)
```

```python
d = {'name': '김경일', 'birth': '0411', 'age': 35}

json_data = json.dumps(d)

json_data
```

```python
json.loads(json_data)
```

```python
d = {"name":"김경일","birth":"0401","age":35}
json_data = json.dumps(d, ensure_ascii=False)
json_data
```

zipfile : 여러개 파일을 zip 형식으로 합치거나 해제

```python
import zipfile

with zipfile.ZipFile('mytext.zip','w') as myzip:
    myzip.write('a.txt')
    myzip.write('b.txt')
    myzip.write('c.txt')
```

```python
with zipfile.ZipFile('mytext.zip') as myzip:
    myzip.extractall()
```
